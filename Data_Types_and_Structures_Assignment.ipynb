# Part 1 – Theoretical Questions

### Q1. What are data structures, and why are they important?

**Answer:**

A **data structure** is a way of organising and storing data so it can be accessed and modified efficiently. Python's built-in structures are **list, tuple, set, dictionary** (plus strings).

**Why important:**
- Choose the right structure → faster, simpler code (e.g. a set for fast membership tests)
- Better memory use and scalability
- Make code readable and easier to maintain
- Foundation for algorithms (searching, sorting, graphs, caching)

### Q2. Explain the difference between mutable and immutable data types with examples.

**Answer:**

- **Mutable** objects can be changed in place after creation: `list`, `dict`, `set`, `bytearray`.
- **Immutable** objects cannot be changed; any 'change' creates a new object: `int`, `float`, `str`, `tuple`, `frozenset`, `bool`, `bytes`.

Immutable objects are hashable (usable as dict keys), thread-safe and safe to share; mutable ones are flexible but need care when shared or passed to functions.

```python
l = [1, 2]; l[0] = 9      # OK
t = (1, 2); t[0] = 9      # TypeError
```

### Q3. What are the main differences between lists and tuples in Python?

**Answer:**

| | List | Tuple |
|---|---|---|
| Syntax | `[1, 2, 3]` | `(1, 2, 3)` |
| Mutability | Mutable | Immutable |
| Methods | Many (`append`, `remove`, `sort`, ...) | Few (`count`, `index`) |
| Speed / memory | Slightly slower, more memory | Faster, less memory |
| Hashable | No | Yes (if elements are hashable) → can be dict keys / set members |
| Typical use | Collections that change | Fixed records, coordinates, return values |

### Q4. Describe how dictionaries store data.

**Answer:**

A dictionary stores **key → value pairs**. Internally it is a **hash table**: Python computes `hash(key)`, uses it to find a slot in an underlying array, and stores the key, value and hash there. Lookups compare the hash and then the key. Keys must be **hashable** (immutable) and unique; values can be anything. Since Python 3.7 dictionaries **preserve insertion order**.

### Q5. Why might you use a set instead of a list in Python?

**Answer:**

- **Uniqueness:** a set automatically removes duplicates.
- **Fast membership test:** `x in set` is O(1) on average vs O(n) for a list.
- **Set algebra:** union `|`, intersection `&`, difference `-`, symmetric difference `^`.

Use a list when you need order, indexing or duplicates.

### Q6. What is a string in Python, and how is it different from a list?

**Answer:**

A **string** (`str`) is an immutable sequence of Unicode characters. A **list** is a mutable sequence of any objects.

| | String | List |
|---|---|---|
| Elements | Characters only | Any type |
| Mutability | Immutable | Mutable |
| Literal | `"abc"` | `[1, 'a']` |
| Methods | `upper`, `split`, `replace`, `find`, ... | `append`, `sort`, `pop`, ... |

Both support indexing, slicing, iteration and `len()`.

### Q7. How do tuples ensure data integrity in Python?

**Answer:**

Tuples are **immutable**: items cannot be added, removed or reassigned after creation. This prevents accidental modification, makes them safe to pass around, usable as **dictionary keys** and set members (if all elements are hashable), and signals to readers that the data is a fixed record. (Caveat: a tuple holding a mutable object, like a list, can still have that inner object modified.)

### Q8. What is a hash table, and how does it relate to dictionaries in Python?

**Answer:**

A **hash table** maps keys to values using a **hash function** that converts a key into an array index, giving **average O(1)** insert, lookup and delete. Collisions (two keys with the same slot) are resolved by probing. Python's `dict` (and `set`) are implemented as hash tables, which is why keys must be hashable and why lookups by key are so fast.

### Q9. Can lists contain different data types in Python?

**Answer:**

Yes. Lists are **heterogeneous**: one list can hold integers, strings, floats, other lists, dictionaries, even functions, e.g. `[1, "hello", 3.5, [2, 3], {"a": 1}]`.

### Q10. Explain why strings are immutable in Python.

**Answer:**

Reasons:
- **Hashability:** strings are used as dict keys, so their hash must never change.
- **Safety and thread-safety:** shared strings cannot be altered unexpectedly.
- **Optimisation:** Python can **intern/cache** identical strings and reuse them, saving memory.
- **Security/predictability:** values such as file paths or names stay constant.

To 'modify' a string, create a new one (`s = s.replace(...)`).

### Q11. What advantages do dictionaries offer over lists for certain tasks?

**Answer:**

- **Fast lookup by key** (O(1) average) instead of scanning (O(n)).
- **Meaningful labels** (`person["name"]`) instead of positions.
- Natural for **mappings, counting, caching, grouping and JSON-like data**.
- Keys are **unique**, so duplicates are handled automatically.

### Q12. Describe a scenario where using a tuple would be preferable over a list.

**Answer:**

When the data is **fixed and should not change**, for example geographic coordinates `(28.61, 77.20)`, an RGB colour `(255, 0, 0)`, a database record, multiple return values from a function, or a **dictionary key** such as `{(row, col): value}` for a grid. Tuples are also slightly faster and smaller.

### Q13. How do sets handle duplicate values in Python?

**Answer:**

A set stores **only unique elements**. Duplicates are silently ignored when creating the set or adding elements: `{1, 2, 2, 3}` → `{1, 2, 3}`, and `s.add(2)` on a set already containing 2 changes nothing. Uniqueness is decided using hash and equality, so elements must be hashable.

### Q14. How does the "in" keyword work differently for lists and dictionaries?

**Answer:**

- **List:** `x in my_list` checks the **values** by scanning elements one by one → **O(n)**.
- **Dictionary:** `x in my_dict` checks the **keys** using the hash table → **O(1)** average. To check values use `x in my_dict.values()` (O(n)).

### Q15. Can you modify the elements of a tuple? Explain why or why not.

**Answer:**

**No.** Tuples are immutable, so item assignment (`t[0] = 5`) raises `TypeError`; there is no `append` or `remove`. Workarounds: convert to a list, modify, convert back, or build a new tuple by concatenation/slicing. If a tuple contains a mutable object (e.g. a list) that object's contents can still change, but the tuple's references to its items cannot.

### Q16. What is a nested dictionary, and give an example of its use case?

**Answer:**

A **nested dictionary** is a dictionary whose values are themselves dictionaries. It models hierarchical data such as JSON/API responses, configurations, or records:

```python
students = {"s1": {"name": "Asha", "marks": {"math": 90, "sci": 85}},
            "s2": {"name": "Ravi", "marks": {"math": 78, "sci": 88}}}
students["s1"]["marks"]["math"]   # 90
```

### Q17. Describe the time complexity of accessing elements in a dictionary.

**Answer:**

- **Average case:** **O(1)** for lookup, insertion and deletion (hash table).
- **Worst case:** **O(n)** if many keys collide (very rare with Python's good hash function and resizing).
- Iterating over all items is O(n).

### Q18. In what situations are lists preferred over dictionaries?

**Answer:**

- When **order and position** matter and data is accessed by **index**.
- When **duplicates** are needed.
- Simple sequences you iterate over, sort, slice or use as stacks/queues.
- When there are no natural unique keys, or you want lower memory use than a dictionary.

### Q19. Why are dictionaries considered unordered, and how does that affect data retrieval?

**Answer:**

Historically (before Python 3.7) dict ordering was an implementation detail, so dictionaries were called *unordered*. Since **Python 3.7 they preserve insertion order**, but they are still not *positional*: you cannot access items by index, only by **key**, and equality ignores order (`{'a':1,'b':2} == {'b':2,'a':1}`). So retrieval is done with keys (`d[key]`, `d.get(key)`), not by position.

### Q20. Explain the difference between a list and a dictionary in terms of data retrieval.

**Answer:**

| | List | Dictionary |
|---|---|---|
| Access by | Integer **index** (`l[2]`) | **Key** (`d['name']`) |
| Index access time | O(1) | O(1) average (by key) |
| Search by value | O(n) | O(n) (by key: O(1)) |
| Missing item | `IndexError` | `KeyError` (or `d.get(key)` returns None) |
| Ordering | Positional | Insertion-ordered, key-based |
